# H₂ Ground-State Energy and Processor Comparison

This notebook studies the ground-state energy of hydrogen (H₂) near its equilibrium bond length and compares a VQE circuit on two simulated processor layouts.

The project uses a bond length of 0.7414 Å. It compares an exact Hamiltonian reference with an ideal VQE result, then evaluates the circuit on noisy simulations of Processor A and Processor B.

## Method

The H₂ Hamiltonian is represented with four qubits. We computed its exact reference energy by diagonalizing the Hamiltonian and used a UCCSD ansatz for VQE. We then transpiled the circuit for each processor’s coupling map and evaluated the ideal VQE parameters with the configured noisy simulator.

The noisy energies were not re-optimized under noise. The comparison therefore measures the effect of applying noise to the ideal VQE circuit and parameters.


```python
import csv
from pathlib import Path

csv_path = Path.cwd() / "results" / "tables" / "architecture_metrics.csv"

with csv_path.open(newline="", encoding="utf-8") as file:
    rows = list(csv.DictReader(file))

for row in rows:
    print(
        f"Processor {row['processor']}: "
        f"noisy energy = {float(row['noisy_energy_hartree']):.6f} Hartree, "
        f"absolute error = {float(row['absolute_error_hartree']):.6f} Hartree, "
        f"depth = {row['depth']}, CX gates = {row['cx_gates']}"
    )
```

## Results

| Processor | Noisy energy (Hartree) | Absolute error (Hartree) | Depth | CX gates |
|---|---:|---:|---:|---:|
| A | -0.904377 | 0.232902 | 91 | 56 |
| B | -0.904377 | 0.232902 | 91 | 56 |

The exact reference energy is **-1.137279 Hartree**. Both noisy estimates are less negative than the reference by 0.232902 Hartree.

![Noisy H₂ energy comparison for Processors A and B](results/figures/energy_comparison.png)

## Interpretation

Both layouts produced the same result because the four-qubit circuit fits the same connected chain on each processor. Their transpiled circuits had the same depth and CX-gate count, so the additional qubits on Processor B did not change this run’s outcome.
